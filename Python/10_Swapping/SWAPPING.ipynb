# Swapping variable

**Swapping** means exchanging the values of two variables.

```python
shiva = 25
shiva1 = 35
```

 1. Swapping Using a Third Variable

```python
temp = shiva
shiva = shiva1
shiva1 = temp
print(shiva,shiva1)
```

```python
pandey = 45
pandey1 = 55
```

```python
temp = pandey
pandey = pandey1
pandey1 = temp
print(pandey)
print(pandey1)
```

2. Swapping Using Arithmetic Operators

```python
sub1 = 15
sub2 = 25 
```

```python
sub1 = sub1+sub2
sub2 = sub1-sub2
sub1 = sub1-sub2
print(sub1)
print(sub2)
```

```python
shiva = 25
shiva1 = 85
```

```python
shiva = shiva+shiva1
shiva1 = shiva-shiva1
shiva = shiva-shiva1
print(shiva,shiva1)
```

3. Swapping Without a Third Variable


```python
shiva1 = 1000
shiva2 = 2000
```

```python
shiva1,shiva2 = shiva2,shiva1
print(shiva1)
print(shiva2)
```

```python
sub = 5000
sub1 = 10000
```

```python
sub,sub1 = sub1,sub
print(sub,sub1)
```
